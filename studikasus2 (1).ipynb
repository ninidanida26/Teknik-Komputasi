# Soal: Analisis Tekanan Atmosfer Bumi Menggunakan Metode Simpson 1/3

Atmosfer Bumi memiliki perubahan tekanan terhadap ketinggian yang dapat dimodelkan dengan persamaan:

$$
P(h)=P_0 \exp\left(-\int_0^h \frac{g}{R\,T(h')}dh'\right)
$$

dengan:

- $P(h)$ = tekanan atmosfer pada ketinggian tertentu (Pa)
- $P_0$ = tekanan atmosfer di permukaan Bumi (Pa)
- $g$ = percepatan gravitasi Bumi (m/s²)
- $R$ = konstanta gas spesifik atmosfer Bumi (J/kg.K)
- $T(h)$ = suhu atmosfer terhadap ketinggian (K)


## Diketahui:

$$
P_0 = 610 \, Pa
$$

$$
g = 3.71 \, m/s^2
$$

$$
R = 188.9 \, J/(kg.K)
$$


Data suhu atmosfer Mars terhadap ketinggian:

| Ketinggian $h$ (km) | Suhu $T(h)$ (K) |
|---|---|
| 0 | 220 |
| 2 | 215 |
| 4 | 210 |
| 6 | 205 |
| 8 | 200 |


Hitung tekanan atmosfer Bumi pada ketinggian:

$$
h = 8 \, km
$$

dengan menggunakan metode integrasi numerik **Simpson 1/3**.


## Integral yang harus dihitung:

$$
I=\int_0^{8000}\frac{g}{R\,T(h)}dh
$$


## Langkah Penyelesaian:

### 1. Tentukan nilai fungsi:

$$
f(h)=\frac{g}{R\,T(h)}
$$


### 2. Hitung integral menggunakan rumus Simpson 1/3:

$$
I \approx \frac{\Delta h}{3}
(f_0+4f_1+2f_2+4f_3+f_4)
$$


### 3. Hitung tekanan atmosfer:

$$
P(8000)=P_0e^{-I}
$$


## Tentukan:

1. Nilai fungsi $f(h)$ pada setiap titik.
2. Nilai integral menggunakan metode Simpson 1/3.
3. Nilai tekanan atmosfer Mars pada ketinggian 8 km.

# Penyelesaian Soal

Diketahui:

$$
P_0 = 610 \ Pa
$$

$$
g = 3.71 \ m/s^2
$$

$$
R = 188.9 \ J/(kg.K)
$$


Data:

| h (m) | T (K) |
|---|---|
| 0 | 220 |
| 2000 | 215 |
| 4000 | 210 |
| 6000 | 205 |
| 8000 | 200 |


## 1. Menentukan fungsi f(h)

Persamaan:

$$
f(h)=\frac{g}{R\,T(h)}
$$


Untuk mempermudah:

$$
\frac{g}{R}=\frac{3.71}{188.9}
$$

$$
=0.01964
$$


Maka:

$$
f(h)=\frac{0.01964}{T(h)}
$$


Hasil perhitungan:

| h (m) | T(K) | f(h) |
|---|---|---|
|0|220|8.93 × 10⁻⁵|
|2000|215|9.14 × 10⁻⁵|
|4000|210|9.35 × 10⁻⁵|
|6000|205|9.58 × 10⁻⁵|
|8000|200|9.82 × 10⁻⁵|


## 2. Menghitung integral Simpson 1/3

Rumus Simpson 1/3:

$$
I=\frac{\Delta h}{3}
(f_0+4f_1+2f_2+4f_3+f_4)
$$


Dengan:

$$
\Delta h = 2000 \ m
$$


Substitusi:

$$
I=
\frac{2000}{3}
[
(8.93\times10^{-5})
+4(9.14\times10^{-5})
+2(9.35\times10^{-5})
+4(9.58\times10^{-5})
+(9.82\times10^{-5})
]
$$


Bagian dalam kurung:

$$
=1.123\times10^{-3}
$$


Maka:

$$
I=
\frac{2000}{3}
(1.123\times10^{-3})
$$


Sehingga:

$$
I=0.7487
$$


## 3. Menghitung tekanan atmosfer Mars pada 8 km


Persamaan:

$$
P(8000)=P_0e^{-I}
$$


Substitusi:

$$
P(8000)=610e^{-0.7487}
$$


Karena:

$$
e^{-0.7487}=0.4729
$$


maka:


$$
P(8000)=610(0.4729)
$$


Hasil:


$$
P(8000)=288.5 \ Pa
$$


# Kesimpulan

Berdasarkan perhitungan menggunakan metode Simpson 1/3, diperoleh:

$$
\boxed{I=0.7487}
$$


Tekanan atmosfer Mars pada ketinggian 8 km adalah:


$$
\boxed{P(8000)=288.5 \ Pa}
$$


Artinya, tekanan atmosfer Bumi mengalami penurunan dari 610 Pa di permukaan menjadi sekitar 288.5 Pa pada ketinggian 8 km.

```python
import numpy as np
import matplotlib.pyplot as plt

# Data suhu Bumi
h = np.array([0, 2000, 4000, 6000, 8000])  # meter
T = np.array([220, 215, 210, 205, 200])    # Kelvin

# Konstanta
g = 3.71       # m/s²
R = 188.9      # J/(kg.K)
P0 = 610       # Pa

# Fungsi integran
f = g/(R*T)

# Integral Simpson 1/3
dh = h[1] - h[0]

I = (dh/3)*(f[0] + 4*f[1] + 2*f[2] + 4*f[3] + f[4])

# Tekanan pada 8 km
P8 = P0*np.exp(-I)

print(f"Integral Simpson = {I:.6f}")
print(f"Tekanan pada 8 km = {P8:.2f} Pa")

# =====================================
# Menghitung tekanan pada tiap titik
# =====================================
P = [P0]

for i in range(1, len(h)):
    delta_I = (g/(R*((T[i-1]+T[i])/2))) * (h[i]-h[i-1])
    P.append(P[-1]*np.exp(-delta_I))

P = np.array(P)

# =====================================
# Grafik Suhu
# =====================================
plt.figure(figsize=(8,5))
plt.plot(h/1000, T, marker='o')
plt.xlabel('Ketinggian (km)')
plt.ylabel('Suhu (K)')
plt.title('Profil Suhu Atmosfer Bumi')
plt.grid(True)
plt.show()

# =====================================
# Grafik Tekanan
# =====================================
plt.figure(figsize=(8,5))
plt.plot(h/1000, P, marker='s')
plt.xlabel('Ketinggian (km)')
plt.ylabel('Tekanan (Pa)')
plt.title('Profil Tekanan Atmosfer Bumi')
plt.grid(True)
plt.show()
```

### Pembuktian Jika Suhu pada Kedua Metode Disamakan

Untuk membuktikan bahwa perbedaan hasil berasal dari penggunaan data suhu yang berbeda, dilakukan perhitungan dengan suhu konstan yang sama pada kedua metode, yaitu:

$$
T = 210 \, K
$$

Maka fungsi yang diintegralkan menjadi:

$$
f(h)=\frac{g}{RT}
$$

$$
f(h)=\frac{3.71}{188.9 \times 210}
$$

$$
f(h)=0.00009335
$$

Karena suhu konstan, maka nilai fungsi pada setiap titik ketinggian juga konstan.

#### Metode Matematis

$$
I=\int_0^{8000}0.00009335 \, dh
$$

$$
I=0.00009335(8000)
$$

$$
I=0.7468
$$

#### Metode Simpson 1/3

Dengan interval:

$$
h=\frac{8000-0}{4}=2000 \, m
$$

Karena semua nilai fungsi sama:

$$
f_0=f_1=f_2=f_3=f_4=0.00009335
$$

Maka:

$$
I=\frac{h}{3}
\left(
f_0+4f_1+2f_2+4f_3+f_4
\right)
$$

$$
I=\frac{2000}{3}
\left(
0.00009335+4(0.00009335)+2(0.00009335)+4(0.00009335)+0.00009335
\right)
$$

$$
I=\frac{2000}{3}(12 \times 0.00009335)
$$

$$
I=0.7468
$$

Terlihat bahwa:

$$
I_{Matematis}=I_{Simpson}=0.7468
$$

Selanjutnya tekanan pada ketinggian 8 km adalah:

$$
P=610e^{-0.7468}
$$

$$
P=289.0 \, Pa
$$

Hasil perhitungan menunjukkan bahwa ketika suhu yang digunakan pada metode matematis dan metode Simpson 1/3 disamakan (konstan sebesar 210 K), kedua metode menghasilkan nilai integral dan tekanan yang sama. Oleh karena itu, perbedaan hasil pada studi kasus ini bukan disebabkan oleh metode perhitungannya, melainkan oleh penggunaan data suhu yang berbeda. Metode matematis menggunakan asumsi suhu konstan, sedangkan metode Simpson 1/3 menggunakan data suhu yang berubah terhadap ketinggian sehingga lebih merepresentasikan kondisi atmosfer yang sebenarnya.
