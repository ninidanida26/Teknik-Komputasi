### No 1
Metode Persegi


```python
import numpy as np

def f(x):
    return 5 + 2*x**2

a = 0
b = 4

prev = 0
n = 1

while True:
    dx = (b-a)/n

    total = 0
    for i in range(n):
        x_mid = a + (i + 0.5)*dx
        total += f(x_mid)

    current = dx * total

    if abs(current - prev) < 1e-14:
        break

    prev = current
    n += 1


print("Nilai integral =", current)
print("n optimal =", n)
```

### No 2
Metode Trapesium

```python
import numpy as np

def C(T):
    return 20 + 0.05*T**2

a = 0
b = 100

prev = 0
n = 1

while True:
    dT = (b-a)/n

    total = C(a) + C(b)

    for i in range(1, n):
        total += 2*C(a + i*dT)

    current = (dT/2) * total

    if abs(current - prev) < 1e-14:
        break

    prev = current
    n += 1

print("Nilai integral =", current)
print("n optimal =", n)
```

# No 3
Metode Simpson

```python
import numpy as np

def I(t):
    return 2 + 3*t + t**2

a = 0
b = 6

prev = 0
n = 2

while True:

    if n % 2 != 0:
        n += 1

    dt = (b-a)/n

    total = I(a) + I(b)

    for i in range(1, n):
        x = a + i*dt

        if i % 2 == 0:
            total += 2*I(x)
        else:
            total += 4*I(x)

    current = (dt/3) * total

    if abs(current - prev) < 1e-14:
        break

    prev = current
    n += 2

print("Nilai integral =", current)
print("n optimal =", n)
```
