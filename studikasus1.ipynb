# Soal Fisika — Metode Bisection pada Medan Listrik


Dua muatan titik ditempatkan pada sumbu x sebagai berikut:

| Muatan | Posisi           | Nilai Muatan   |
| ------ | ---------------- | -------------- |
| (q_1)  | (x=-1m) | (+4  C)
| (q_2)  | (x=3 m)  | (-1  C)

Gunakan konstanta Coulomb:

k = 8.99 × 10^9 N m²/C²

Medan listrik total pada suatu titik x di sumbu x diberikan oleh:



```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# KONSTANTA DAN MUATAN
# ==========================================

k = 8.99e9

q1 = 4      # C
q2 = -1     # C

x1 = -1     # m
x2 = 3      # m

# ==========================================
# FUNGSI MEDAN LISTRIK TOTAL
# ==========================================

def E(x):

    E1 = k*q1*(x - x1)/abs(x - x1)**3
    E2 = k*q2*(x - x2)/abs(x - x2)**3

    return E1 + E2

# ==========================================
# METODE BISECTION
# ==========================================

a = 5
b = 10

toleransi = 1e-6
iterasi_maks = 100

if E(a)*E(b) > 0:
    print("Interval tidak memenuhi syarat bisection")
else:

    for i in range(iterasi_maks):

        c = (a + b)/2

        if abs(E(c)) < toleransi:
            break

        if E(a)*E(c) < 0:
            b = c
        else:
            a = c

    print("Akar ditemukan:")
    print("x =", c)

# ==========================================
# VISUALISASI
# ==========================================

x = np.linspace(-5, 12, 1000)
y = [E(i) for i in x]

plt.figure(figsize=(8,6))

plt.plot(x, y, label='E(x)')
plt.axhline(0, color='black')

# posisi muatan
plt.plot(x1, 0, 'bo', markersize=10, label='q1')
plt.plot(x2, 0, 'ro', markersize=10, label='q2')

# titik akar
plt.plot(c, 0, 'go', markersize=12, label='E = 0')

plt.xlabel('x [m]')
plt.ylabel('E(x)')

plt.title('Metode Bisection pada Medan Listrik')

plt.grid(True)
plt.legend()

plt.show()
```

# LAPORAN

**1. Pendahuluan**

   Medan listrik merupakan salah satu konsep penting dalam fisika yang menjelaskan pengaruh gaya listrik di sekitar muatan listrik. Konsep ini didasarkan pada Hukum Coulomb yang menyatakan bahwa gaya listrik antara dua muatan berbanding lurus dengan besar muatan dan berbanding terbalik dengan kuadrat jarak antara keduanya. Pada sistem yang terdiri dari lebih dari satu muatan, medan listrik total diperoleh dengan menjumlahkan kontribusi medan listrik dari setiap muatan.

   Dalam banyak kasus, persamaan medan listrik yang terbentuk bersifat non-linear sehingga sulit diselesaikan secara analitik. Oleh karena itu, digunakan metode numerik seperti metode bisection untuk mencari solusi pendekatan, yaitu menentukan titik ketika medan listrik total bernilai nol. Dengan bantuan pemrograman seperti Python, proses perhitungan dan visualisasi medan listrik dapat dilakukan dengan lebih mudah sehingga hasil analisis menjadi lebih jelas dan akurat.



**2. Metodologi**

   Penelitian ini dilakukan dengan pendekatan teori dan komputasi untuk menentukan titik di mana medan listrik total bernilai nol akibat dua muatan titik. Langkah pertama adalah menentukan parameter fisika yang digunakan, yaitu besar muatan, posisi masing-masing muatan, serta konstanta Coulomb. Selanjutnya, medan listrik pada suatu titik di sumbu x dihitung berdasarkan Hukum Coulomb dengan mempertimbangkan kontribusi dari setiap muatan menggunakan prinsip superposisi.

   Karena persamaan medan listrik yang diperoleh bersifat non-linear, maka penyelesaiannya dilakukan menggunakan metode numerik bisection. Metode ini bekerja dengan menentukan interval awal yang memiliki perubahan tanda fungsi, kemudian membagi interval tersebut secara bertahap hingga diperoleh nilai akar dengan tingkat ketelitian yang diinginkan. Proses perhitungan dan visualisasi dilakukan menggunakan bahasa pemrograman Python untuk membantu analisis serta menggambarkan grafik medan listrik secara lebih jelas.

**3. Hasil dan Pembahasan**



  Titik medan listrik nol terjadi ketika besar medan listrik dari kedua muatan sama:

$$
E_1 = E_2
$$

Rumus medan listrik:

$$
E = k\frac{q}{r^2}
$$

Karena titik yang dicari berada di sebelah kanan muatan kedua $(x > 3)$, maka:

- Jarak terhadap $q_1$:

$$
r_1 = x+1
$$

- Jarak terhadap $q_2$:

$$
r_2 = x-3
$$

Substitusi ke persamaan medan listrik:

$$
k\frac{4}{(x+1)^2}
=
k\frac{1}{(x-3)^2}
$$

Konstanta Coulomb saling menghilangkan:

$$
\frac{4}{(x+1)^2}
=
\frac{1}{(x-3)^2}
$$

Kalikan silang:

$$
4(x-3)^2 = (x+1)^2
$$

Kembangkan persamaan:

$$
4(x^2 - 6x + 9)=x^2+2x+1
$$

$$
4x^2 -24x +36 = x^2+2x+1
$$

Pindahkan semua ruas ke kiri:

$$
3x^2 -26x +35 =0
$$
